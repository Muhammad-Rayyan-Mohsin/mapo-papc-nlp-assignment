# MAPO + PAPC — Inference Demo

Load the trained LoRA adapter and answer a multilingual math problem.

Run from the repo root so relative paths resolve:

```bash
jupyter notebook notebooks/01_inference_demo.ipynb
```

````python
import os, sys, yaml
if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')
sys.path.insert(0, os.getcwd())
from src.model import load_policy, generate_answer
cfg = yaml.safe_load(open('config.yaml'))
print('base model:', cfg['models']['base_model'])
````

````python
model, tokenizer = load_policy(
    base_model=cfg['models']['base_model'],
    adapter_path='checkpoints/mapo-adapter',
)
print('policy loaded')
````

## English

````python
problem_en = "Miguel uses 2 pads of paper a week for his drawing. If there are 30 sheets of paper on a pad of paper, how many sheets of paper does he use every month?"
print(generate_answer(model, tokenizer, problem_en, cfg['generation']))
````

## Swahili (low-resource — PAPC target language)

````python
problem_sw = "Miguel hutumia pedi 2 za karatasi kwa wiki katika uchoraji wake. Ikiwa kuna kurasa 30 za karatasi kwenye pedi moja ya karatasi, huwa anatumia kurasa ngapi za karatasi kila mwezi?"
print(generate_answer(model, tokenizer, problem_sw, cfg['generation']))
````
